Interpretation of Statistics

```python
###Mean: represents the average value of the data.

###Median:This helps to understand central tendency without being affected by outliers.

###Mode: This indicates the most frequently occurring values.

###Standard_Deviation : This shows how much the values vary from the mean.
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Histogram

```python
plt.figure(figsize=(6,3))
plt.title("Histogram for numerical column")
plt.xlabel("Volume")
plt.ylabel("Frequency")
plt.hist(data["Volume"],color='green',bins=10)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.title("Histogram for numerical column")
plt.xlabel("Avg Price")
plt.ylabel("Frequency")
plt.hist(data["Avg Price"],color='midnightblue',bins=10)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.title("Histogram for numerical column")
plt.xlabel("Total Sales Value")
plt.ylabel("Frequency")
plt.hist(data["Total Sales Value"],color='red',bins=10)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.title("Histogram for numerical column")
plt.xlabel("Discount Rate (%)")
plt.ylabel("Frequency")
plt.hist(data["Discount Rate (%)"],color='brown',bins=10)
plt.show()
```

```python
plt.figure(figsize=(6,3))
plt.title("Histogram for numerical column")
plt.xlabel("Net Sales Value")
plt.ylabel("Frequency")
plt.hist(data["Net Sales Value"],color='pink',bins=10)
plt.show()
```

```python
###Inferences:
#From the histograms, we can observe how the data values are distributed. Any skewness indicates imbalance in the data, and extreme bars may suggest outliers.
```

```python
#Inferece:
#The box plot shows the spread and distribution of numerical variables such as Sales, Discount, Profit, etc.

#The middle line inside the box represents the median, which indicates the central value of the data.

#The height of the box (Interquartile Range – IQR) shows where the middle 50% of the data lies.

#Whiskers indicate the normal range of values excluding outliers.

#Points outside the whiskers are outliers, representing unusually high or low values.
```
