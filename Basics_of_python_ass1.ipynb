 Exercise 1: Prime_Numbers

```python
def prime_no(num):
    if num <= 1:
        print("Not exist")
    elif num == 2:
        print("Prime number")
    elif num % 2 == 0:
        print("Not a prime number")
    else:
        print("Prime number")

prime_no(2)
```

Exercise 4: Hospital Billing System

```python
Room_charges_per_day=2000*4
Doctor_consultation_fee=1500
Lab_test_charges=3*300
Medicine_charges=2400

total_bill=Room_charges_per_day+Doctor_consultation_fee+Lab_test_charges+Medicine_charges

if total_bill>10000:
    discount=total_bill*0.1
    print("discount is applied")
else:
    discount=0
    print("discount is not applied")
    
final_bill=total_bill-discount

print("total_bill",total_bill)
print("discount",discount)
print("final bill result",final_bill)
    
```
