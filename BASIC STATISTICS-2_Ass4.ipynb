Problem Statement : Hospital Patient Data Analysis

1.	Load the patient dataset 

```python
import pandas as pd
import numpy as np

billing=pd.read_csv("Billing_Data.csv")
billing
```

```python
patient=pd.read_csv("Patient_Data.csv")
patient
```

```python
patient.info()
```

2.	Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].

```python
billing_col=['PatientID', 'Department', 'Doctor', 'BillAmount']
patient_billing_col=patient[billing_col]
```

```python
patient_billing_col.head()
```

3.	Drop administrative columns like ['ReceptionistID', 'CheckInTime'].

```python
drop_col=patient.drop(columns=['ReceptionistID', 'CheckInTime'],axis=1,inplace=True)
drop_col
```

```python
patient
```

4.	Use groupby to find total bill amount per department.

```python
total_bill=patient.groupby('Department')['BillAmount'].sum()
total_bill
```

5.	Remove duplicate patient records based on PatientID.

```python
patient_dup=patient.drop_duplicates('PatientID')
```

```python
patient_dup
```

6.	Fill missing BillAmount values with the mean bill amount

```python
bill_amt_values=patient['BillAmount'].fillna(patient['BillAmount'].mean())
bill_amt_values
```

7.	Merge the billing dataset with patient dataset on PatientID.

```python
merge_set=pd.merge(patient,billing,on="PatientID",how='inner')
merge_set
```

8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).

```python
newpatients = pd.DataFrame({
    'PatientID': [106, 107],
    'Name': ['Ram', 'Sam'],
    'Department': ['Dermatology', 'Cardiology'],
    'Doctor': ['Dr.Will', 'Dr.Fig'],
    'BillAmount': [6000, 7000],
    'InsuranceCovered': [3000, 2500],
    'FinalAmount': [3500, 4000]
})

concat_set=pd.concat([merge_set,newpatients],ignore_index=True)
concat_set
```

9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).

```python
df = pd.DataFrame({
    'BillID': [101, 102, 103],
    'TotalAmount': [5000, 3000, 7000]
})
new_cols = pd.DataFrame({
    'InsuranceCovered': [2000, 1000, 3000],
    'FinalAmount': [3000, 2000, 4000]
})

df_final = pd.concat([df, new_cols], axis=1)
```
