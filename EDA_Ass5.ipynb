BAR CHART

```python
category_map = {1: 'Normal', 2: 'Suspect', 3: 'Pathological'}
cardio['Category'] = cardio['NSP'].round().map(category_map)
category_counts = cardio['Category'].value_counts()
```

```python
plt.figure(figsize=(6,4))
plt.bar(category_counts.index, category_counts.values, color='skyblue')
plt.xlabel('Category')
plt.ylabel('Count')
plt.title('Frequency of Categories (Bar Chart)')
plt.show()
```

```python
#Insight:

#The bar chart shows that the Normal category has the highest frequency.

#Suspect and Pathological cases occur less frequently, indicating class imbalance.

#This imbalance should be considered during predictive modeling to avoid biased results.
```

SCATTER PLOT

```python
plt.figure(figsize=(6,4))
plt.scatter(cardio['LB'], cardio['ASTV'], alpha=0.6)
plt.xlabel('LB')
plt.ylabel('ASTV')
plt.title('Scatter Plot: LB vs ASTV')
plt.show()
```

```python
#Insight:

#The scatter plot indicates a weak to moderate relationship between baseline heart rate and short-term variability.

#No clear linear trend is observed, suggesting that LB alone is not sufficient to predict ASTV.

#Clusters of points may indicate different fetal health states.
```

```python
corr=cardio.corr(numeric_only=True)
corr
```

```python
plt.figure(figsize=(12,4))
sns.heatmap(corr, annot=True, cmap='vlag',fmt='.1f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#Insight:

#Strong correlations are observed among variability-related features such as ASTV, MSTV, ALTV, and MLTV.

#Baseline FHR (LB) shows weaker correlation with most other variables.

#Highly correlated features may be redundant and can be reduced using feature selection techniques.
```

PAIRPLOT

```python
sns.pairplot(cardio[['LB', 'ASTV', 'MSTV', 'Width', 'NSP']],kind='hist',hue='NSP')
plt.show()
```

```python
#Insights:

#Pair plots reveal clear separation patterns between NSP categories for variability-based features.

#Normal cases tend to cluster tightly, while pathological cases show greater spread.

#This indicates that variability features are strong predictors of fetal condition.
```
